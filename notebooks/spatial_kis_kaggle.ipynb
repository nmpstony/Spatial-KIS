# Spatial KIS preprocessing
Attach two private Kaggle Datasets: one containing this `spatial-kis` source folder and one containing the MP4 corpus. Set the two paths below, enable GPU, then Run All. The completed artifact is written to `/kaggle/working` and can be saved as a Kaggle Output version.

```python
from pathlib import Path
import os, shutil
SOURCE_DATASET = Path('/kaggle/input/spatial-kis-source/spatial-kis')
VIDEO_DATASET = Path('/kaggle/input/private-videos')
WORKSPACE = Path('/kaggle/working/spatial-kis')
ARTIFACT = Path('/kaggle/working/spatial-kis-artifact')
if not WORKSPACE.exists():
    shutil.copytree(SOURCE_DATASET, WORKSPACE)
os.environ['KAGGLE_INPUT_PATH'] = str(VIDEO_DATASET)
os.environ['KAGGLE_OUTPUT_PATH'] = str(ARTIFACT)
print(VIDEO_DATASET, ARTIFACT)
```
